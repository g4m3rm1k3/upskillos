# Machine Learning Is Old Math: the map

Almost nothing in machine learning is new mathematics. The "algorithms" are old, proven ideas, mostly from the 1600s-1900s, that computers finally made cheap enough to run on huge amounts of data. If the maths feels like the hard part, that's because it is the *whole subject*: once you can see which old idea an algorithm is, the code is short.

This series teaches the **maths first, with the code alongside it**, in the order the ideas depend on each other. Every lesson follows the same shape:

1. **The old idea**: who figured it out, when, and what problem they were trying to solve. (Not trivia: knowing the original problem tells you what the idea is *for*.)
2. **What it becomes in ML**: the modern algorithm it turned into.
3. **Small steps**: each step explains one thing, you type a few lines, run them, and compare with the expected output. You derive formulas by hand, then check them in code (the habit that catches your own mistakes).
4. **Guided exploration**: named alternatives to swap in (e.g. absolute error instead of squared error; momentum instead of plain descent), on a copy, then return to the original.
5. **A challenge**, with the worked solution at the bottom of the notebook.

**How to run a notebook:** upload it to Google Colab (File → Upload notebook). Below each code snippet is an empty code cell: type the snippet in yourself rather than pasting, run it, then check the output shown after. Where a class is being built across steps, later steps tell you to go back and add a method to the same cell.

## Reading the notation

Textbooks hide simple ideas behind symbols. A quick key to what you'll meet in Lessons 1-4 (each is explained again when it first appears):

| Symbol | Read it as | In code |
|---|---|---|
| `f(x)` | "the function f applied to x" | `f(x)` |
| `f'(x)` or `df/dx` | slope of f at x | `slope_by_wiggling(f, x)` |
| `f''(x)` | slope of the slope (curvature) | `second_slope_by_wiggling(f, x)` |
| `∂f/∂x` | slope with respect to x, holding other inputs still | `partial_derivative(f, point, index)` |
| `∇f` | the list of all the partial slopes: the gradient | `gradient(f, point)` |
| `Σᵢ` | "add up the following for every item i" | `sum(...)` or a `for` loop with a running total |
| `x̄` | the mean of x | `mean(x_values)` |
| `≈` | approximately equal | |
| `lim h→0` | make h smaller and smaller, see where the answer settles | the shrinking-gap loop |
